```python
import json
import numpy as np
import torch
json_path = '/home/om/mpi/data-gen/genzoo/outputs/horse_dataset/train.json'

with open(json_path, 'r') as f:
    data = json.load(f)
data_points = data['data']

for sample in data_points:
    kp_3d = torch.tensor(sample['keypoint_3d']).reshape(1,-1,3)  # Reshape to (1, 1, 3) for a single sample
    kp_2d = torch.tensor(sample['keypoint_2d']).reshape(1,-1,3)[:, :, :2]  # Reshape to (1, 1, 2) for a single sample
    camera_params = torch.tensor(sample['trans']).reshape(1,3)
    
    # Assuming camera_params contains 'focal_length' and 'principal_point'
    focal_length = torch.tensor([[1000,1000]])  # Example focal length, replace with actual value if available
    # Project 3D keypoints to 2D
    projected_kp_2d = perspective_projection(kp_3d, camera_params,focal_length)
    
    # Compare projected 2D keypoints with the provided 2D keypoints
    error += np.linalg.norm(np.array(projected_kp_2d) - np.array(kp_2d))


print("Average projection error:", error / len(data_points))
```
